# Gaia wide-binary constraints on alpha * eta (phenomenological)

This notebook downloads the public El-Badry Gaia EDR3 wide-binary catalog, computes per-system conservative 1-sigma upper limits on the product `alpha * eta` using the phenomenological mapping `dv/v ~ 0.5 * (alpha * eta)`, and produces summary plots.

Assumptions:
- Total system mass default = 2 M_sun (adjustable).
- Projected separation used as an estimator for orbital radius (circular orbit approximation).
- Proper-motion errors provide transverse-velocity uncertainty; radial-velocity uncertainties are not included.
- Filter for high-quality parallaxes and RUWE where available.

```python
import os
import requests
from astropy.table import Table
import numpy as np
import matplotlib.pyplot as plt
from math import sqrt

# Constants
G = 6.67430e-11  # m^3 kg^-1 s^-2
M_sun = 1.98847e30  # kg
AU = 1.495978707e11  # m
pc = 3.085677581e16  # m
km = 1e3

# El-Badry dataverse download URL (Harvard Dataverse)
url = 'https://dataverse.harvard.edu/api/access/datafile/4661493?format=original&gbrecs=true'
fname = 'elbadry_widebinaries_edr3_all_columns.fits.gz'

if not os.path.exists(fname):
    print('Downloading El-Badry catalog (this may take a while)...')
    r = requests.get(url, stream=True)
    r.raise_for_status()
    with open(fname, 'wb') as f:
        for chunk in r.iter_content(chunk_size=8192):
            if chunk:
                f.write(chunk)
    print('Download complete:', fname)
else:
    print('Found cached file:', fname)
```

```python
# Read the FITS file with astropy and inspect column names
tbl = Table.read(fname)
print('Rows:', len(tbl))
print('Columns:', tbl.colnames)

# Common column name heuristics used in El-Badry catalogs
# Try to identify columns for primary/secondary ra/dec/parallax/pmra/pmdec/errors and projected separation
cols = tbl.colnames
candidates = {}
for name in ['ra_1','ra_2','dec_1','dec_2','parallax_1','parallax_2','parallax','parallax_error_1','parallax_error_2','pmra_1','pmdec_1','pmra_error_1','pmdec_error_1','pmra_error_2','pmdec_error_2','pmra','pmdec','pmra_error','pmdec_error','s','sep','r_proj','r_proj_au','rproj']:
    if name in cols:
        candidates[name] = True
print('Detected candidate columns (subset):', list(candidates.keys()))
```

```python
# Heuristic mapping based on common El-Badry naming conventions. Adjust if the table differs.
# Many El-Badry outputs use columns like ra_1, dec_1, ra_2, dec_2, parallax_1, parallax_2, pmra_1, pmdec_1, etc.

# Define accessors with fallbacks
def get_col(tbl, names):
    for n in names:
        if n in tbl.colnames:
            return tbl[n]
    return None

ra1 = get_col(tbl, ['ra_1','ra1','ra_1_'])
dec1 = get_col(tbl, ['dec_1','dec1','dec_1_'])
ra2 = get_col(tbl, ['ra_2','ra2'])
dec2 = get_col(tbl, ['dec_2','dec2'])
par1 = get_col(tbl, ['parallax_1','parallax1','parallax_1_'])
par2 = get_col(tbl, ['parallax_2','parallax2'])
par = get_col(tbl, ['parallax'])
par_err1 = get_col(tbl, ['parallax_error_1','parallax_error1','parallax_error'])
pmra1 = get_col(tbl, ['pmra_1','pmra1','pmra_1_','pmra_1_x'])
pmdec1 = get_col(tbl, ['pmdec_1','pmdec1','pmdec_1_','pmdec_1_x'])
pmra_err1 = get_col(tbl, ['pmra_error_1','pmra_error1','pmra_err_1','pmra_err'])
pmdec_err1 = get_col(tbl, ['pmdec_error_1','pmdec_error1','pmdec_err_1','pmdec_err'])
sep_arcsec = get_col(tbl, ['s','sep','sep_arcsec','r_ang'])
rproj_au = get_col(tbl, ['r_proj_au','rproj_au','r_proj_au_','r_proj'])

print('Found: ra1', ra1 is not None, 'dec1', dec1 is not None, 'parallax', par is not None, 'sep_arcsec', sep_arcsec is not None, 'rproj_au', rproj_au is not None)
```

```python
# Compute distance (pc) and projected separation (m). Use parallax_1 if available, else parallax average.
if par1 is not None:
    dist1_pc = 1.0 / (par1)  # parallax in mas? El-Badry likely uses mas; if arcsec adjust accordingly
    # If parallax is in mas, convert: distance_pc = 1000/parallax_mas. Detect scale by median dist heuristics.
    # Heuristic: if median(par) > 1e-3 assume parallax in mas.
    median_par = np.nanmedian(par1)
    if median_par > 1e-3:
        dist1_pc = 1000.0 / par1
    else:
        dist1_pc = 1.0 / par1
elif par is not None:
    if np.nanmedian(par) > 1e-3:
        dist1_pc = 1000.0 / par
    else:
        dist1_pc = 1.0 / par
else:
    # fallback: use a fixed distance for computation -> 100 pc
    print('No parallax columns found; using default 100 pc for distance estimation.')
    dist1_pc = np.full(len(tbl), 100.0)

# projected separation: prefer rproj_au, else compute from angular sep and distance
if rproj_au is not None:
    r_proj_m = np.array(rproj_au, dtype=float) * AU
elif sep_arcsec is not None:
    # convert arcsec to radians and multiply by distance in pc -> projected separation in AU approx: sep_arcsec * dist_pc = sep_AU
    r_proj_au = np.array(sep_arcsec, dtype=float) * np.array(dist1_pc)
    r_proj_m = r_proj_au * AU
else:
    # As last fallback, compute angular sep from RA/Dec columns if present
    if ra1 is not None and dec1 is not None and ra2 is not None and dec2 is not None:
        import astropy.coordinates as coord
        import astropy.units as u
        c1 = coord.SkyCoord(ra=np.array(ra1)*u.deg, dec=np.array(dec1)*u.deg)
        c2 = coord.SkyCoord(ra=np.array(ra2)*u.deg, dec=np.array(dec2)*u.deg)
        sep = c1.separation(c2).arcsec
        r_proj_au = sep * np.array(dist1_pc)
        r_proj_m = r_proj_au * AU
    else:
        raise RuntimeError('Cannot determine projected separation: no suitable columns found. Inspect tbl.colnames and adjust mapping.')

# compute transverse proper-motion error (mas/yr) from pmra_err and pmdec_err if available, else use pmra_error columns
if pmra_err1 is not None and pmdec_err1 is not None:
    pmerr_masyr = np.sqrt(np.array(pmra_err1, dtype=float)**2 + np.array(pmdec_err1, dtype=float)**2)
else:
    # fallback: try combined pm error column
n    pmerr_masyr = None

# if no pm error available, use a conservative default (0.02 mas/yr)
if pmerr_masyr is None:
    print('No pm error columns found; using default sigma_mu = 0.02 mas/yr')
    pmerr_masyr = np.full(len(tbl), 0.02)

# distance in pc as float array
dist_pc = np.array(dist1_pc, dtype=float)

# transverse velocity uncertainty in km/s: sigma_v = 4.74047 * sigma_mu(mas/yr) * d(pc) / 1e3 (to convert m/s?)
# 4.74047 * mu(mas/yr) * d(pc) gives km/s directly
sigma_v_kms = 4.74047 * pmerr_masyr * dist_pc

# orbital speed in km/s for total mass M_tot (kg) at radius r_proj_m
M_tot = 2.0 * M_sun  # default total mass; adjust as needed
v_kms = np.sqrt(G * M_tot / r_proj_m) / 1000.0

# conservative 1-sigma upper limit on alpha*eta per system: alpha_eta < 2 * (sigma_v / v)
# avoid divide by zero or tiny v -> mask invalid
mask = np.isfinite(v_kms) & (v_kms > 1e-6)
alpha_eta_limits = np.full(len(tbl), np.nan)
alpha_eta_limits[mask] = 2.0 * (sigma_v_kms[mask] / v_kms[mask])

# Clip unreasonably large limits for plotting (keep them but note)

print('Computed alpha*eta limits for', np.sum(mask), 'systems (masked invalid entries).')
```

```python
# Summary statistics and plots
valid = ~np.isnan(alpha_eta_limits)
vals = alpha_eta_limits[valid]
print('alpha*eta percentiles (10,50,90):', np.nanpercentile(vals, [10,50,90]))

plt.figure(figsize=(8,4))
plt.hist(vals, bins=np.logspace(-6,2,80), histtype='stepfilled')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('1-sigma upper limit on alpha * eta')
plt.ylabel('Count')
plt.title('Distribution of conservative alpha*eta limits from El-Badry catalog (phenomenological)')
plt.grid(True, which='both', ls=':')
plt.show()

# show sample of best systems (smallest alpha*eta limits)
idx = np.argsort(vals)[:10]
print('Top 10 systems (smallest limits):')
for i in idx:
    print(f'limit={vals[i]:.3e}, v_kms={v_kms[valid][i]:.3e} km/s, sigma_v={sigma_v_kms[valid][i]:.3e} km/s')
```

## Notes and next steps
- The notebook uses heuristics to find columns; if the FITS table uses different names you may need to adjust the `get_col` name lists.
- The result bounds are conservative and per-system; for a population constraint you can combine systems statistically (inverse-variance weighting) or select a high-quality subsample.
- To translate alpha*eta -> alpha you need estimates of eta for stars; that requires theoretical calculation or the laboratory Bi-209 calibration.
